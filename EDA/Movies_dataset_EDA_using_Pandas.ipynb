**Spearman Correlation Concept**

```python
# Positive correlation - sugar increases then chocolate price increases (or sugar decreses chocolate price dcreases) i.e same effect
# Negative correlation - petrol price increases car demand decreses or vice versa i.e oppositive effect
# No correlation - price of phone and price of milk - no correlation
# value lies between -1 to 1
# value closer to 0 like 0.22 or -0.12 then it is weak correlation
# value like 0.88 --> strong positive correlation and if value like -0.91 --> strong negative correlation
# we can take threshold value as 0.5 and -0.5 to determine string or weak correlation if nothing is given regarding threshold
```

**Which Director has the highest "Hit Rate"? (Percentage of movies rated > 7.5)**


```python
# overall percentage of movies rated > 7.5
x=df["id"].count()
a=df[(df["rating"] > 7.5)]
y=a["id"].count()
result=(y/x)*100
round(result,2)
```

```python
total_movies = df.groupby("directors_name")["id"].count()
hit_movies = df[(df["rating"] > 7.5)].groupby("directors_name")["id"].count()
hit_rate = (hit_movies / total_movies * 100).fillna(0)     # numpy array supports element wise division as count column is a series
result = hit_rate.sort_values(ascending=False).head(1)
result

# fillna(0) is used because some directors may have 0 hit movies (no movies with rating > 7.5).
# it ensures that directors with no hit movies are included with a hit rate of 0% instead of NaN.
```

```python
# By value counts

total_movies = df["directors_name"].value_counts()
hit_movies = df[(df["rating"] > 7.5)]
hit=hit_movies["directors_name"].value_counts()
hit_rate = (hit / total_movies * 100).fillna(0)
result = hit_rate.sort_values(ascending=False).head(5)
result
```

**Difference between groupby and value_counts**

For pure frequency counting, value_counts() is usually shorter and more readable.
For anything more complex (averages, sums, multiple columns, custom calculations), use groupby().

value_count : When to use -->
Count occurrences of values in one column.
Quick frequency analysis.
Simpler and shorter syntax.

groupby : When to use -->
Sum, mean, min, max, count, etc.
Group by multiple columns.
Perform custom aggregations.

**Analyse the "Sequel Effect" — Do movies with numbers in titles earn more?**

```python
tile_with_numbers = df[(df["name"].str.contains(r"\d", na=False))]  # in case of no title present give false
title_without_numbers = df[~(df["name"].str.contains(r"\d", na=False))]

mean_income_movies_with_numbers = tile_with_numbers["gross_income"].mean()
mean_income_movies_without_numbers = title_without_numbers["gross_income"].mean()

print("Sequel Effect : ",(True if (mean_income_movies_with_numbers > mean_income_movies_without_numbers) else False))
```

**Do movies with more actors (ensemble cast) earn more?**

```python
multiple_actor_movies = df[(df["stars_name"].str.contains(r",", na=False))]  # in case of no stars present give false
single_actor_movies = df[~(df["stars_name"].str.contains(r",", na=False))]

mean_income_multiple_actor_movies = multiple_actor_movies["gross_income"].mean()
mean_income_single_actor_movies = single_actor_movies["gross_income"].mean()

print("Do movies with more actors earn more : ",(True if (mean_income_multiple_actor_movies > mean_income_single_actor_movies) else False))
```
